Найти k-ый по величине элемент массива. 

Важно: нельзя просто отсортировать массив. 

Для решения необходимо использовать minheap. 

На выходе должно быть две реализции:

1) без использования heapq (кучу и методы работы с ней реализовываем сами)

2) с использованием heapq 

```python
def find_kth_largest_no_heapq(nums, k):
    # Создаём пустую кучу (список)
    heap = []

    def heapify_up(idx):
        while idx > 0:
            parent = (idx - 1) // 2
            if heap[idx] < heap[parent]:
                heap[idx], heap[parent] = heap[parent], heap[idx]
                idx = parent
            else:
                break

    def heapify_down(idx):
        size = len(heap)
        while True:
            left = 2 * idx + 1
            right = 2 * idx + 2
            smallest = idx

            if left < size and heap[left] < heap[smallest]:
                smallest = left
            if right < size and heap[right] < heap[smallest]:
                smallest = right
            if smallest == idx:
                break
            
            heap[idx], heap[smallest] = heap[smallest], heap[idx]
            idx = smallest

    def push(val):
        heap.append(val)
        heapify_up(len(heap) - 1)

    def pop():
        if len(heap) == 1:
            return heap.pop()
        root = heap[0]
        heap[0] = heap.pop()
        heapify_down(0)
        return root

    for num in nums:
        if len(heap) < k:
            push(num)
        elif num > heap[0]:  # heap[0] — минимальный элемент
            pop()
            push(num)

    return heap[0]  # k-й по величине элемент
```

```python
import heapq

def find_kth_largest_with_heapq(nums, k):
    heap = []
    for num in nums:
        if len(heap) < k:
            heapq.heappush(heap, num)
        elif num > heap[0]:
            heapq.heappop(heap)
            heapq.heappush(heap, num)
    return heap[0]
```

# Тесты

```python
# Базовые случаи (из задания)
print(find_kth_largest_no_heapq([3, 2, 1, 5, 6, 4], 2) == 5)
print(find_kth_largest_with_heapq([3, 2, 1, 5, 6, 4], 2) == 5)
# k = 1 (максимальный элемент)
print(find_kth_largest_no_heapq([1, 2, 3, 4, 5], 1) == 5)
print(find_kth_largest_with_heapq([1, 2, 3, 4, 5], 1) == 5)
# k = len(nums) (минимальный элемент)
print(find_kth_largest_no_heapq([5, 4, 3, 2, 1], 5) == 1)
print(find_kth_largest_with_heapq([5, 4, 3, 2, 1], 5) == 1)
# Повторяющиеся элементы
print(find_kth_largest_no_heapq([3, 3, 3, 3, 3], 3) == 3)
print(find_kth_largest_with_heapq([3, 3, 3, 3, 3], 3) == 3)
# Массив с отрицательными числами
print(find_kth_largest_no_heapq([-1, -2, -3, -4], 2) == -2)
print(find_kth_largest_with_heapq([-1, -2, -3, -4], 2) == -2)
# Массив из одного элемента
print(find_kth_largest_no_heapq([42], 1) == 42)
print(find_kth_largest_with_heapq([42], 1) == 42)
# Случайные данные (несколько вариантов)
print(find_kth_largest_no_heapq([10, 7, 11, 5, 13, 8], 3) == 10)
print(find_kth_largest_with_heapq([10, 7, 11, 5, 13, 8], 3) == 10)
print(find_kth_largest_no_heapq([99, 1, 100, 2, 98], 2) == 99)
print(find_kth_largest_with_heapq([99, 1, 100, 2, 98], 2) == 99)
print(find_kth_largest_no_heapq([2, 8, 7, 1, 3, 5, 6, 4], 5) == 4)
print(find_kth_largest_with_heapq([2, 8, 7, 1, 3, 5, 6, 4], 5) == 4)
# 
```
